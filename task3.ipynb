Problem Statement 1: Employee Performance Bonus Eligibility

```python
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# Find the highest score
max_score = max(employees.values())

# Find all employees with the highest score
top_performers = []
for name, score in employees.items():
    if score == max_score:
        top_performers.append(name)

# Display the results
print(f"Top Performers Eligible for Bonus: {', '.join(top_performers)} (Score: {max_score})")
```

Problem Statement 2: Search Query Keyword Analysis


```python
import string

query = "Buy mobile phone buy phone online"
query = query.lower()
query = query.translate(str.maketrans('', '', string.punctuation))
words = query.split()

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1
result = {}

for word, count in frequency.items():
    if count > 1:
        result[word] = count

print(result)
```

Problem Statement 4: Email Domain Usage Analysis


```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

count_gmail = 0
count_yahoo = 0

# Count emails by domain
for mail in emails:
    if mail.endswith('gmail.com'):  # Check if email ends with gmail.com
        count_gmail += 1  # Increment counter (not append!)
    elif mail.endswith('yahoo.com'):  # Check if email ends with yahoo.com
        count_yahoo += 1

# Calculate total emails
total_emails = len(emails)

# Calculate percentages
gmail_percentage = (count_gmail / total_emails) * 100
yahoo_percentage = (count_yahoo / total_emails) * 100

# Display results
print(f"gmail.com: {gmail_percentage:.0f}%")
print(f"yahoo.com: {yahoo_percentage:.0f}%")
```

Problem Statement 5: Sales Spike Detection


```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
average_sales = sum(sales) / len(sales)

# Option 2: Use a slightly lower threshold (1.29 instead of 1.3)
threshold = average_sales * 1.29  # 1700 * 1.29 = 2193

print("Sales Spikes Detected (Day, Sales Value):")
for day, sales_value in enumerate(sales, start=1):
    if sales_value > threshold:
        print(f"Day {day}: {sales_value}")
```

Problem Statement 6: Duplicate User ID Detection

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

frequency = {}

for user in user_ids:
    frequency[user] = frequency.get(user, 0) + 1

for user, count in frequency.items():
    if count > 1:
        print(f"{user} → {count} times")
```
