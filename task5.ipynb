Problem 1: Social Media – Post Engagement Analyzer

```python
def analyze_post_engagement(likes_list):
    total_likes = 0


    for likes in likes_list:
        total_likes += likes


    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)



likes = [200, 150, 300, 250, 150]
analyze_post_engagement(likes)
```

Problem 3: Agriculture – Rainfall Adequacy Checker


```python
def rainfall_checker(rainfall_list, required_level):
    total = 0

    # Loop to calculate total rainfall
    for rain in rainfall_list:
        total += rain

    average = total / len(rainfall_list)

    # Condition check
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", average)
    print("Rainfall Status:", status)



rainfall_data = [70, 75, 68, 80, 67]
rainfall_checker(rainfall_data, 70)
```

Problem 4: Social Media – Duplicate Account Detection

```python
def detect_duplicate_accounts(usernames):

    unique_users = set(usernames)

    if len(unique_users) != len(usernames):
        result = "Yes"
    else:
        result = "No"

    print("Duplicate Accounts Found:", result)

users = ["rahul123", "priya123", "rahul123", "arjun789"]
detect_duplicate_accounts(users)
```

Problem 9: E-Learning – Student Attendance Eligibility System


```python
def check_exam_eligibility(attendance_list):
    total_classes = len(attendance_list)
    present_count = 0

    for status in attendance_list:
        if status == "P":
            present_count += 1

    attendance_percentage = (present_count / total_classes) * 100

    if attendance_percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"

    print("Attendance Percentage:", attendance_percentage)
    print("Exam Eligibility:", eligibility)


attendance_data = ["P", "P", "A", "P", "P"]
check_exam_eligibility(attendance_data)
```

Problem 10: Smart Electricity Bill Analyzer

```python
def calculate_electricity_bill(units):
    bill = 0

    # Slab Calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7

    # Usage Classification
    if bill < 500:
        usage_status = "Low Usage"
    elif bill <= 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"

    return bill, usage_status

total_bill, usage = calculate_electricity_bill(250)

print("Total Bill:", total_bill)
print("Usage Level:", usage)
```
