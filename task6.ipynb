Problem Statement 1:Smart Parking Lot Management System


```python
def smart_parking_lot_management(capacity, vehicle_logs):

    current_parked = 0
    peak_usage = 0

    for log in vehicle_logs:
        if log == "IN":
            current_parked += 1
        elif log == "OUT":
            current_parked -= 1

        if current_parked > peak_usage:
            peak_usage = current_parked

    if current_parked > capacity:
        status = "Over Capacity - Alert!"
    else:
        status = "Available"

    return current_parked, peak_usage, status
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

current, peak, status = smart_parking_lot_management(capacity, vehicle_logs)

print("Currently Parked Vehicles:", current)
print("Peak Parking Usage:", peak)
print("Parking Status:", status)
```

 Problem Statement 2:Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic_level, weather):

    speed = 20
    base_time = (distance / speed) * 60

    if traffic_level == "Low":
        traffic_delay = 5
    elif traffic_level == "Medium":
        traffic_delay = 10
    else:
        traffic_delay = 20

    if weather == "Clear":
        weather_delay = 0
    elif weather == "Rainy":
        weather_delay = 15
    else:
        weather_delay = 25


    final_eta = base_time + traffic_delay + weather_delay

    return round(final_eta)

distance = 8
traffic = "High"
weather = "Rainy"

eta = estimate_delivery_time(distance, traffic, weather)

print("Estimated Delivery Time:", eta, "minutes")
```

Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

```python
def movie_theater_seat_occupancy_analyzer(booked_seats):
    total_seats = 200

    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    elif occupancy >= 50:
        status = "Moderate Occupancy"
    else:
        status = "Seats Available"

    return occupancy, status

booked_seats = [1] * 150

occupancy, status = movie_theater_seat_occupancy_analyzer(booked_seats)

print("Occupancy:", occupancy, "%")
print("Show Status:", status)


```

 Problem Statement 4:Cloud Server Load Classification System

```python
def cloud_server_load(classification_list):
  # Calculate Average_CPU once for the entire list
  Average_CPU = sum(classification_list) / len(classification_list)
  status = "" # Initialize status
  if Average_CPU <= 50:
    status="Normal"
  elif Average_CPU <=80:
    status="Warning"
  else:
    status="Critical"
  return Average_CPU, status # Return both values

classification_list=[45,60,70,85,90]
avg_cpu_result, server_status_result = cloud_server_load(classification_list)
print("Average CPU load:", avg_cpu_result)
print("Server status:", server_status_result)
```

 Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def smart_classroom_resource_monitor(resource_usage):
    overused_resources = []

    for resource, hour in resource_usage.items():
        if hour > 6:   # Threshold limit
            overused_resources.append(resource)

    if overused_resources:
        energy_alert = "Yes"
    else:
        energy_alert = "No"

    print("Overused Resources:", ", ".join(overused_resources))
    print("Energy Alert:", energy_alert)

smart_classroom = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

smart_classroom_resource_monitor(smart_classroom)
```

 Problem Statement 6: Online Event Registration Capacity Controller

```python
def event_registration_capacity_controller(registrations):
    event_capacity = 100

    if registrations <= event_capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = event_capacity
        waitlisted = registrations - event_capacity
        status = "Closed"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

registrations = int(input("Enter number of registrations: "))
event_registration_capacity_controller(registrations)
```
