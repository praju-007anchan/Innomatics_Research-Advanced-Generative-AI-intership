```python
# 2. Pass / Fail Analyzer

Student_mark=[45,78,90,33,60]
pass_student=0
failed_student=0
for i in Student_mark:
    if i >=50:
       pass_student+=1
    else:
        failed_student+=1
        
        
print(f"student passed:{pass_student}") 
print(f"student failed:{failed_student}") 
   
```

```python
#3. Simple Data Cleaner
names = [" Alice ", "bob", " CHARLIE "]
Correct_name=[]
for name in names:
    Correct_name.append(name.strip().lower())

print(Correct_name)    


```

```python
#4.# Message Length Analyzer

messages = ["Hi", "Welcome to the platform", "OK"]
short_count = 0
long_count =0 

for message in messages:
    if len(message) < 10:
        short_count +=1          
    else:
         long_count+=1
         print (f"message size is too long:{message}")
         
print(f"Number of short messages: {short_count}")
print(f"Number of long messages: {long_count}")         
         
         
```

```python
#5. Error Message Detector

logs = ["INFO", "ERROR", "WARNING", "ERROR"]
ERROR_count=0
for i in logs:
    if i=='ERROR':
        ERROR_count+=1
print(ERROR_count)
```
