Problem 1: Smart Light Controller


```python
# CLASS
class SmartLight:

    def __init__(self, name):
        self.name = name
        self.status = "OFF"

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def display_status(self):
        print(self.name, "is", self.status)

light_name = input("Enter Light Name: ")
action = input("Enter Action (ON/OFF): ").upper()   # Convert to uppercase

# OBJECT
light = SmartLight(light_name)

# METHOD CALL
if action == "ON":
    light.turn_on()
elif action == "OFF":
    light.turn_off()
else:
    print("Invalid Action")

light.display_status()
```

Problem 2: Employee ID Card System

```python

class Employee:

    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_info(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


# Taking input from user
name = input("Employee Name: ")
emp_id = input("Employee ID: ")
department = input("Department: ")

# Creating object
emp1 = Employee(name, emp_id, department)

# Displaying ID card
emp1.display_info()
```

Problem 3: Mobile Contact Record


```python
class Mobile_record:

    def __init__(self, name, number):
        self.name = name
        self.number = number

    def display_info(self):
        print("Contact Saved")
        print("Name:", self.name)
        print("Phone:", self.number)


# Taking input
contact_name = input("Contact Name: ")
phone_number = input("Phone Number: ")

# Validation: Number should be exactly 10 digits
if phone_number.isdigit() and len(phone_number) == 10:
    contact1 = Mobile_record(contact_name, phone_number)
    contact1.display_info()
else:
    print("Invalid phone number! Must be exactly 10 digits.")
```

Problem 4: Product Price Tag Generator

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.price = price   # price stored as integer

    def display_price_tag(self):
        print("Product:", self.name)
        print("Price: ₹", self.price)


# Taking input from user
product_name = input("Product Name: ")
price = int(input("Price: "))   # converting to integer

# Creating object
product1 = Product(product_name, price)

# Displaying price tag
product1.display_price_tag()
```

Problem 5: Movie Rating Display System


```python
class Movie:

    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_movie(self):
        print("Movie:", self.name)
        print("Rating:", self.rating, "/ 10")


# Taking input
movie_name = input("Movie Name: ")
rating = float(input("Rating: "))

# Validation: Rating must be between 0 and 10
if 0 <= rating <= 10:
    movie1 = Movie(movie_name, rating)
    movie1.display_movie()
else:
    print("Invalid rating! Rating must be between 0 and 10.")
```

Problem 6: Delivery Address Manager

```python
class Delivery:

    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def display_delivery(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# Taking input
customer_name = input("Customer Name: ")
address = input("Address: ")

# Creating object
delivery1 = Delivery(customer_name, address)

# Displaying details
delivery1.display_delivery()
```
