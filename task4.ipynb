Problem 1: Mobile Recharge Validation System


```python
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount: ₹"))

        if amount < 50:
            print(" Recharge amount must be at least ₹50.")
        elif amount not in valid_plans:
            print(" Invalid plan. Available plans:", valid_plans)
        else:
            print(" Recharge successful!")
            break
validate_recharge()
```

Problem 2: Inventory Reorder Alert System


```python
alter_system = {"chair": 20, "Door": 15, "Wood": 10, "books": 8}

def warehouse(inventory):
    for product, stock in inventory.items():
        if stock < 15:
            print(f"{product}: Reorder Alert")
        else:  # stock >= 15
            print(f"{product}: Stock OK")

# Call the function
warehouse(alter_system)

```

Problem 3: Student Result Processing System

```python
student = [40, 50, 60, 10]

def process_student_marks(marks):
    total = 0
    for mark in marks:
        total += mark
    average = total / len(marks)
    if average >= 50:
        print("Pass")
    else:
        print("Fail")
    return average
avg = process_student_marks(student)
print(f"Average marks: {avg}")
```


Problem 4: Cab Fare Estimator with Retry Option


```python
def calculate_fare():
    base_fare = 50
    rate_per_km = 12

    while True:
        distance = float(input("Enter distance in km: "))
        peak = input("Is it peak hour? (yes/no): ").lower()

        fare = base_fare + (distance * rate_per_km)

        if peak == "yes":
            fare += fare * 0.25

        print(f"Estimated Fare: ₹{fare}")

        retry = input("Do you want to calculate again? (yes/no): ").lower()
        if retry != "yes":
            break
calculate_fare()
```

Problem 5: Employee Attendance Eligibility Checker

```python
def check_eligibility(attendance_list):
    present_count = 0

    for day in attendance_list:
        if day == "P":
            present_count += 1

    percentage = (present_count / len(attendance_list)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]
print("Eligibility:", check_eligibility(attendance))
```

Problem 6: Password Strength Checker

```python
def check_password_strength(password):
    if len(password) < 8:
        return False

    has_digit = False
    has_special = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
        elif ch in "@#$":
            has_special = True

    return has_digit and has_special
pwd = input("Enter password: ")
if check_password_strength(pwd):
    print("Strong")
else:
    print("Weak")
```
