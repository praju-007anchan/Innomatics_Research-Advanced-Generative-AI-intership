Problem 1: Unique Words in a Sentence

```python
sentence = input("Enter sentence: ")
words = sentence.split() #sting>List
unique_words = set(words)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)

```

Problem 2: Highest Salary from Employee Data


```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_name = ""
highest_salary = 0

for name, salary in employees.items():
    if salary > highest_salary:
        highest_salary = salary
        highest_name = name

print(f"Highest Salary: {highest_name} - {highest_salary}")


```

Problem 3: Find Maximum and Minimum Values

```python
numbers = [45, 22, 89, 10, 66]
maximum_numbers=numbers[0]
minimum_numbers=numbers[0]
for num in numbers:
    if num > maximum_numbers:
        maximum_numbers = num
    if num < minimum_numbers:
        minimum_numbers = num

print("List:", numbers)
print("Max:", maximum_numbers)
print("Min:", minimum_numbers)
```

Problem 4: Count Products Above a Price Threshold


```python
prices=[450,1200,899,1500,300]
count=0 # count how many price is above 1000
for price in prices:
    if price > 1000:
        count +=1
print("Product above 1000:", count)
```

Problem 5: Calculate Attendance Percentage


```python
attendance = ["P", "P", "A", "P", "P"]
present=0
total_day=len(attendance)
for p in attendance:
  if p=="P":
    present+=1
percentage = (present / total_day) * 100
print(f"Attendance Percentage: {percentage:.2f}%")

```



Problem 7: Count Character Frequency


```python
text = "PPTHON"

text = text.lower()

char_count = {}
for char in text:
    char_count[char] = char_count.get(char, 0) + 1
print(char_count)
```
